# SPARQL-VI OG-RAG Testing

**Two-Ontology Architecture**: DSL Hypergraph (Syntax) + Domain Hypergraph (Schema)

**Full English Ontology** for optimal retrieval effectiveness

## 1. Setup & Load Engine

```python
import yaml
import importlib
import sys

# Remove cached modules to force reload
if 'query_engine.sparql_vi_ograg_engine' in sys.modules:
    del sys.modules['query_engine.sparql_vi_ograg_engine']
if 'query_engine.sparql_vi_hypergraph' in sys.modules:
    del sys.modules['query_engine.sparql_vi_hypergraph']

from query_engine.sparql_vi_ograg_engine import SPARQLVIQueryEngine

# Load API key
with open("api_keys.yaml") as f:
    groq_api_key = yaml.safe_load(f).get("GROQ_API_KEY")

# Initialize engine with FULL ENGLISH ontologies
engine = SPARQLVIQueryEngine(
    dsl_facts_path="data/dsl/sparql_vi/dsl_facts.json",
    domain_facts_path="data/dsl/sparql_vi/domain_facts.json",
    test_cases_path="data/dsl/sparql_vi/test_cases.json",
    llm_api_key=groq_api_key
)

print(f"\n✅ Dual Hypergraphs initialized:")
print(f"   - DSL: {len(engine.dsl_hypergraph.nodes)} nodes, {len(engine.dsl_hypergraph.edges)} edges")
print(f"   - Domain: {len(engine.domain_hypergraph.nodes)} nodes, {len(engine.domain_hypergraph.edges)} edges")
```

## 2A. Check DSL Facts - Query Types Available

```python
test_query = "Find average prices of product which has price > 10000"
print(f"Question: {test_query}\n")

# Retrieve context from BOTH hypergraphs
print("📚 Retrieving context from DUAL hypergraphs...")
context = engine.retrieve_context(
    test_query, 
    dsl_top_k=8,     # INCREASED to 10 to get more query type options
    domain_top_k=8   # Top 15 domain facts (ensure :Person class retrieved)
)

print(f"\n{'='*80}")
print("CONTEXT SENT TO LLM (should be ENGLISH):")
print('='*80)
print(context)
print('='*80)

# Generate query with OG-RAG
print("\n🤖 Generating with OG-RAG...")
vi_result = engine.generate_ograg(
    test_query, 
    dsl_top_k=8,  # Match retrieval parameter
    domain_top_k=8
)
print(f"\nGenerated SPARQL-VI:\n{vi_result}")
```

## 3. Full Evaluation on 30 Test Cases

```python
import json
from datetime import datetime

# Run evaluation on all test cases
print("🚀 Starting full evaluation on 30 test cases...")
print("   Method: OG-RAG with dual hypergraphs\n")

results = engine.evaluate_all(
    method='ograg',
    dsl_top_k=5,
    domain_top_k=7,  # Increased for better retrieval
    nodes_top_k=5
)

# Save results
timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
output_file = f"results/sparql_vi_eval_{timestamp}.json"

with open(output_file, 'w', encoding='utf-8') as f:
    json.dump(results, f, indent=2, ensure_ascii=False)

print(f"\n✅ Results saved to: {output_file}")

# Display summary
print("\n" + "="*80)
print("EVALUATION SUMMARY (OG-RAG with English Ontology)")
print("="*80)
print(f"Total Cases: {results['total_cases']}")
print(f"Syntax Accuracy: {results['syntax_accuracy']:.1%}")
print(f"Exact Match: {results['exact_match_accuracy']:.1%}")
print("="*80)
```

## 4. Analyze Failed Cases

```python
# Analyze failed cases to understand retrieval issues
failed_cases = [r for r in results['results'] if not r['syntax_valid']]

print(f"📊 Total failed cases: {len(failed_cases)}\n")
print("="*80)

for i, case in enumerate(failed_cases[:3], 1):  # Show first 3 failed cases
    print(f"\n🔴 FAILED CASE #{i}: [{case['test_case_id']}]")
    print(f"Question: {case['question']}")
    print(f"Error: {case['errors']}")
    print(f"\nGenerated query:\n{case['generated_vi']}")
    
    # Show retrieved context for this case
    print(f"\n{'─'*80}")
    print("CONTEXT RETRIEVED:")
    print('─'*80)
    context = engine.retrieve_context(
        case['question'], 
        dsl_top_k=5, 
        domain_top_k=15
    )
    print(context[:1500] + "..." if len(context) > 1500 else context)
    print("\n" + "="*80)
```
